```python
# prompt: zip root_dir, excluding the directory Backups; save it to a dated backup folder to root_dir/Backups, creating directory if it doesn't exist

import os
from datetime import datetime
import zipfile

backup_dir = os.path.join(root_dir, 'Backups')
if not os.path.exists(backup_dir):
  os.makedirs(backup_dir)

now = datetime.now()
date_string = now.strftime("%Y-%m-%d_%H-%M-%S")
zip_filename = os.path.join(backup_dir, f"backup_{date_string}.zip")

with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
  for foldername, subfolders, filenames in os.walk(root_dir):
    if 'Backups' not in foldername:
      for filename in filenames:
        filepath = os.path.join(foldername, filename)
        arcname = os.path.relpath(filepath, root_dir)
        zipf.write(filepath, arcname)
```
